# 3. LED matrix

Now that the COM port is open and the controller responds, test the LED matrix. Connect it to the **LED1** port.

The matrix is a 5×5 grid of 25 addressable RGB LEDs. In code it is an array of shape `(25, 3)`: one row per LED, three columns for red, green and blue, each `0`–`255`. LED `0` is the first in the chain. Run the single-LED cell to see which physical corner that is, because it depends on how the matrix is mounted.

This matters more than "make it blink": in a microscope the illumination pattern *is* an experimental parameter. Which LEDs you switch on sets the angle light hits the sample from, which changes contrast and, for the half-lit patterns further down, lets you pull out phase information from an otherwise transparent specimen.

<img src="img//IMG20230926153841.jpg" alt="LED matrix connected to the LED1 port" width="500"/>

````python
# test LED
print("The LED pin is: "+str(ESP32.led.get_ledpin()))
time.sleep(2)
ESP32.led.send_LEDMatrix_full(intensity=(255, 255, 255))
time.sleep(2.5)
ESP32.led.send_LEDMatrix_full(intensity=(0, 0, 0))
````

````python
# display random pattern
for i in range(100):
    led_pattern = np.random.randint(0,55, (25,3))
    ESP32.led.send_LEDMatrix_array(led_pattern=led_pattern,timeout=1)

ESP32.led.send_LEDMatrix_single(indexled=0, intensity=(0, 255, 0))
````

````python
# walk a single lit LED through the matrix, one at a time
# this is how you find out which physical LED is index 0, 1, 2, ...
for repeat in range(5):
    for iLED in range(25):
        ESP32.led.send_LEDMatrix_full(intensity=(0, 0, 0))
        ESP32.led.send_LEDMatrix_single(indexled=iLED, intensity=(255, 255, 255))
        time.sleep(0.1)

ESP32.led.send_LEDMatrix_full(intensity=(0, 0, 0))  # leave the matrix off
````

````python
# half-lit illumination: left half red, right half green
led_pattern = np.zeros((25, 3))
list_left = (0, 1, 2, 3, 4, 5, 9, 10, 11, 12, 13, 14, 15, 16, 17)
list_right = (0, 5, 6, 7, 8, 9, 18, 19, 20, 21, 22, 23, 24)
led_pattern[list_left, 0] = 255   # column 0 = red
led_pattern[list_right, 1] = 255  # column 1 = green
ESP32.led.send_LEDMatrix_array(led_pattern=led_pattern, timeout=1)
time.sleep(1)
ESP32.led.send_LEDMatrix_array(led_pattern=led_pattern * 0, timeout=1)  # all off
````

# 4. Motors

Connect the Z-stage to the correct motor port. **The external power supply must be on.** USB alone cannot drive the motors, and an unpowered motor looks exactly like code that is not sending commands.

Axes are numbered: A = 0, X = 1, Y = 2, Z = 3.

Movement is in **steps**, not millimetres. A step is one increment of the stepper motor, so the physical distance per step depends on your gearing and lead screw. You have to calibrate it yourself: move a known number of steps, measure the actual displacement, and you have your steps-per-mm. Do this before you trust any position value, because every quantitative measurement you make later depends on it.

> **Careful:** there are no limit switches fitted by default, so nothing stops the stage in software. If you command a large move the stage will happily drive into its own end stop and grind. Start with small step counts until you know the travel range.

`speed` is in steps per second. `is_blocking=True` makes Python wait until the move finishes; with `False` the command returns immediately and the motor keeps going in the background, which is what you want when moving several axes at once.

<img src="img//IMG20230926161544.jpg" alt="Motorised Z-stage" width="350"/>

````python
# The boards ship pre-configured, so you normally do NOT need to run this.
# `if 0:` means the block below is switched off - change to `if 1:` only if a TA
# tells you the pin configuration was lost, or you are wiring a different board.
if 0:
    # pin assignment for the WEMOS R32 D1
    ESP32.motor.set_motor(stepperid = 1, position = 0, stepPin = 26, dirPin=16, enablePin=12, maxPos=None, minPos=None, acceleration=None, isEnable=1)
    ESP32.motor.set_motor(stepperid = 2, position = 0, stepPin = 25, dirPin=27, enablePin=12, maxPos=None, minPos=None, acceleration=None, isEnable=1)
    ESP32.motor.set_motor(stepperid = 3, position = 0, stepPin = 17, dirPin=14, enablePin=12, maxPos=None, minPos=None, acceleration=None, isEnable=1)
    ESP32.motor.set_motor(stepperid = 0, position = 0, stepPin = 19, dirPin=18, enablePin=12, maxPos=None, minPos=None, acceleration=None, isEnable=1)
````

The pin configuration can also be reset to default values. For the built-in functions, see the UC2-REST pin configurator notebook: https://github.com/openUC2/UC2-REST/blob/master/DOCUMENTATION/DOC_UC2Client-PinConfigurator.ipynb


````python
# Read position, move the Y axis back and forth at three speeds, read position again.
# Each pair of moves returns to the starting point, so the two printed positions
# should match. If they don't, the motor skipped steps - lower the speed.
position1 = ESP32.motor.get_position(timeout=1)
print("before:", position1)

ESP32.motor.move_y(steps=-10000, speed=5000, is_blocking=True)
ESP32.motor.move_y(steps=10000, speed=5000, is_blocking=True)

ESP32.motor.move_y(steps=-10000, speed=10000, is_blocking=True)
ESP32.motor.move_y(steps=10000, speed=10000, is_blocking=True)

ESP32.motor.move_y(steps=-10000, speed=100000, is_blocking=True)
ESP32.motor.move_y(steps=10000, speed=100000, is_blocking=True)

time.sleep(1) 

position2 = ESP32.motor.get_position(timeout=1)
print("after: ", position2)
````

**Homing** means driving an axis until it hits a known reference point, so the controller can call that position zero. Without it, "position 0" is just wherever the stage happened to be when you powered on, and it changes every session, which makes repeatable coordinates impossible.

This needs a **limit switch**: a small mechanical switch the stage presses at the end of its travel. These are not fitted by default, so you would have to solder one on yourself. Ask a TA if you want to. The command below homes in one direction, and only works if a switch is wired and positioned correctly.

````python
ESP32.home.home_y(speed=20000,direction=-1)
````

The `ESP32` object keeps a connection to the board open until you restart the kernel or unplug the USB. Close it explicitly when you are done, otherwise the COM port stays locked and the next person (or your next kernel) cannot connect.

````python
ESP32.close()
````

## 5.2 ArduCam

The ArduCam is the general-purpose colour camera. The quickest way to look through it is the built-in Windows **Camera** app. You will have to switch cameras, since it defaults to your laptop webcam.

From Python it is an ordinary USB webcam, so plain OpenCV handles it; no vendor SDK needed. The cell below opens a live preview window and saves a frame when you press `s`.

[ArduCam spec sheet](https://www.arducam.com/product/b0196arducam-8mp-1080p-usb-camera-module-1-4-cmos-imx219-mini-uvc-usb2-0-webcam-board-with-1-64ft-0-5m-usb-cable-for-windows-linux-android-and-mac-os/)

<img src="img/arducam.jpg" alt="ArduCam USB camera module" style="width: 250px; margin-right: 20px;">

````python
import cv2

# Index 0 is usually your laptop's built-in webcam. If you see your own face
# instead of the microscope, try 1, then 2, until you get the right camera.
camera_index = 0
vid = cv2.VideoCapture(camera_index)

if not vid.isOpened():
    raise RuntimeError(f"Could not open camera {camera_index} - is it plugged in, "
                       "and is another program (Windows Camera app) using it?")

frame_counter = 0

while True:
    ret, frame = vid.read()
    if not ret:
        print("Lost the camera feed.")
        break

    cv2.imshow('frame', frame)

    key = cv2.waitKey(1) & 0xFF
    if key == ord('s'):                      # 's' saves the current frame
        filename = f"arducam_{frame_counter:03d}.png"
        cv2.imwrite(filename, frame)
        print("saved", filename)
        frame_counter += 1
    elif key == ord('q'):                    # 'q' quits
        break

vid.release()
cv2.destroyAllWindows()

# The preview opens in a separate window, which can hide behind VS Code - check
# your taskbar. The window must have focus for 's' and 'q' to register; closing
# it with the X does not stop the loop, so always quit with 'q'. If the window
# does get stuck, restart the kernel.
````

# 7. Troubleshooting

Work down this list before asking a TA. The reader has the longer version.

- **Restart the kernel** before every reconnection attempt, not as a last resort. It releases the COM port and fixes a surprising amount.
- `PermissionError(13, 'Access is denied.')` — something else is holding the COM port: a previous kernel, the WebSerial test tab, Arduino IDE, or Cura. Close them and restart the kernel.
- `FileNotFoundError(2, ...)` — wrong COM port number, or the USB is not properly connected. Check Device Manager again.
- `ModuleNotFoundError` — wrong kernel, or you skipped the import cell. Check the kernel name in the top right and run the cells in order from the top.
- **Motors do nothing** — external power supply connected and switched on? Motors without power look exactly like software that is not sending commands.
- **Nothing responds at all** — try the [WebSerial test](https://youseetoo.github.io/indexWebSerialTest.html) to separate hardware from software problems, then close that tab so it releases the port.
- **Last resort** — with `8DM020` activated: `pip install UC2-REST --force-reinstall --user`.

If none of this helps, reach out to a TA.
